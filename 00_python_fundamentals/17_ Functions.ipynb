### Python Built-in Functions 
Built-in functions are **pre-defined functions** that come with Python. You can use them without importing any modules. They are always available.

>  **Think of it like:** Built-in functions are like a toolbox that comes with Python – you don't need to buy or build tools; they're already there!

---
###  Input/Output Functions

| Function | Description | Example |
|----------|-------------|---------|
| `print()` | Displays output | `print("Hello")` |
| `input()` | Takes user input | `name = input("Enter name: ")` |

###  Type Conversion Functions

| Function | Description | Example |
|----------|-------------|---------|

### User-Defined Functions
A **function** is a reusable block of code that performs a specific task. It helps you organize code, avoid repetition, and make programs more readable.
Creating a function using `def` keyword

> **Think of it like:** A function is like a recipe – you define the steps once, then you can use it whenever you want to cook that dish!

```python
# Function without any arguments
def greet():      # defining a function named 'greet
    print("Hello! welcome to the program.")

greet()  # Calling the function to execute it   
```

```python
# Function with arguments 
def greet_user(name):   # defining a function named 'greet_user' that takes one argument 'name'
    print(f"Hello {name}! welcome to the program.")

greet_user("Hasib")  # Calling the function with an argument   
```

```python
# Function with return value
def square(x): 
    number = x * x  # local variable to store the square of the number  
    return number


square(5) # Calling the function with an argument and returning the value
```

```python
def check_age(age):
    if age < 0:
        return "Invalid age!"
    if age < 18:
        return "Minor"
    return "Adult"

print(check_age(25))   # Adult
print(check_age(-5))   # Invalid age!
print(check_age(15))   # Minor
```

###  Modifying Global Variables
A Global Variable is a variable that is defined outside of any function (at the top level of a script/module). It can be accessed from anywhere in the program that both inside and outside functions.

```python
counter = 0 # global variable
def increment():
    global counter  # Declare to modify global
    counter += 1

increment() # adding 1 to the global variable 'counter'
increment() # adding 1 to the global variable 'counter'
print(counter)   
```

### Global vs Local Variables

```python
# Global variable
global_var = "I am global"

def my_function():
    # Local variable
    local_var = "I am local"
    print(local_var)   # Accessible inside function
    print(global_var)  # Accessible inside function

my_function()
```

```python
print(local_var) # ❌ Error! Not accessible outside
```

```python
print(global_var)  # Accessible outside
```

### Temperature Converter

```python
def celcius_to_fahrenheit(celcius):    # funcion to conver celcius to fahrenheit
    fahrenheit = (celcius * 9/5) + 30
    return fahrenheit

def fahrenheit_to_celcius(fahrenheit):  # function to convert fahrenheit to celcius 
    celcius = (fahrenheit - 32) * 5/9
    return celcius

print(celcius_to_fahrenheit(25))
```

```python
fahrenheit_to_celcius(77)
```

### BMI Calculator

```python
def calculate_bmi(weight, height): 
    bmi = weight / (height ** 2)
    return bmi

def get_bmi_category(bmi):
    if bmi < 18.5:
        return "Underweight" 
    elif bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obese"

weight = 65 #kg
height = 1.75 #m

bmi = calculate_bmi(weight, height)
category = get_bmi_category(bmi)

print(f"BMI: {bmi:.1f}")
print(f"Category: {category}")
```
