1. View (Shallow Copy)  
A view creates a new reference to the original data without creating a completely independent copy.  
* Both share the same underlying data.
* Changes in one may affect the other.  
* Saves memory.

2. Deep Copy  
A deep copy creates a completely independent copy of the original data.
* Both have separate data.
* Changes in one do not affect the other.
* Uses more memory.

```
arr = np.array([[1, 2, 3], [4, 5, 6]])

# These create VIEWS (shallow copy)
view1 = arr[:]           # Slice
view2 = arr[1:, :]        # Slice
view3 = arr.ravel()       # Ravel (usually)
view4 = arr.T             # Transpose

# These create COPIES (deep copy)
copy1 = arr.copy()        # Explicit copy
copy2 = arr.reshape(3, 2)  # Reshape (usually)
copy3 = arr.flatten()      # Flatten

# Check with base attribute
print(view1.base is arr)   # True (view)
print(copy1.base is None)  # True (copy)


````python
import numpy as np
````

````python
arr = np.array([1, 2, 3, 4, 5])

# Create a deep copy
c = np.copy(arr)  # or arr.copy()
c
````

````python
# Modify the c
c[0] = 99
print(c)
print(arr) #Original unchanged!
````

````python
# Check if it's independent
print(c.base is None) # True (independent)
````

### Practical Example

````python
# WRONG - Creates view
original = np.array([1, 2, 3, 4, 5])
modified = original[1:4]  # View!
modified
````

````python
modified[0] = 99
print(original)  # [1 99 3 4 5] - Oops! Original changed
````

````python
# CORRECT - Create copy
original = np.array([1, 2, 3, 4, 5])
modified = original[1:4].copy() #copy
modified[0] = 99

print(modified)   # [99 3 4]
print(original)  # [1 2 3 4 5] - Original safe
````

>When in doubt, use .copy() to ensure data safety!
