### Broadcasting  
Broadcasting allows NumPy to perform operations on arrays of different shapes.  

Rules
1. Arrays are aligned from the right
2. Dimensions must match or be 1
3. Missing dimensions are treated as 1




```python
import numpy as np
```

Scalar broadcasting

```python
# Create a 1D array
a = np.array([1, 2, 3])

# Add a scalar (which gets broadcasted to [2, 2, 2])
result = a + 2

print("Original array:", a)
print("After adding scalar 2:", result)
```

Row vector broadcasting

```python
# Create a 2D matrix (shape: 2x3)
matrix = np.array([[1, 2, 3], [4, 5, 6]])

# Create a row vector (shape: 3,)
row_vector = np.array([10, 20, 30])

# Add row_vector to matrix (shape compatibility: (2, 3) + (3,) works!)
result = matrix + row_vector

print("Original matrix:\n", matrix)
print("Row vector to add:\n", row_vector)
print("Result of broadcasting:\n", result)
```

Column vector broadcasting

```python
# Create a 2D matrix (shape: 2x3)
matrix = np.array([[1, 2, 3], [4, 5, 6]])

# Create a column vector (shape: 2x1)
col_vector = np.array([[10], [20]])

# Add col_vector to matrix (shape compatibility: (2, 3) + (2, 1) works!)
result = matrix + col_vector

print("Original matrix:\n", matrix)
print("Column vector to add:\n", col_vector)
print("Result of broadcasting:\n", result)
```
