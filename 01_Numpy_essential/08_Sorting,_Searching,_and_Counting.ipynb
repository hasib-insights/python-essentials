# Sorting, Searching, and Counting

### 1. Sorting
**Sorting** means arranging the elements of a NumPy array in a specific order (ascending by default).
* **Key Function**: `np.sort(array)`
* **Example**: Arranging `np.array([3, 1, 2])` becomes `[1, 2, 3]`.

### 2. Searching
**Searching** means finding the indexes or elements in a NumPy array that match a specific condition.
* **Key Function**: `np.where(condition)`
* **Example**: Finding where elements are greater than 5 in an array.

### 3. Counting
**Counting** means determining the number of elements that meet a condition, or finding the frequencies of unique values.
* **Key Functions**: `np.sum(condition)` (counts `True` values) or `np.unique(array, return_counts=True)`
* **Example**: Counting how many elements in `np.array([1, 2, 3, 2])` are equal to `2` (result is 2).
---

```python
import numpy as np
```

## Sorting (returns a copy by default)

```python
arr = np.array([3, 1, 4, 1, 5, 9, 2, 6])

print(np.sort(arr))
```

```python
idx = np.argsort(arr)
print(idx)
```

### `np.argsort()` Tabular Map

How the original indices map to the sorted values:

| Sorted Position | Target Value | Found at Original Index | Original Array Alignment |
| :---: | :---: | :---: | :--- |
| 1st | 1 | 1 | `arr[1] = 1` |
| 2nd | 1 | 3 | `arr[3] = 1` |
| 3rd | 2 | 6 | `arr[6] = 2` |
| 4th | 3 | 0 | `arr[0] = 3` |
| 5th | 4 | 2 | `arr[2] = 4` |
| 6th | 5 | 4 | `arr[4] = 5` |
| 7th | 6 | 7 | `arr[7] = 6` |
| 8th | 9 | 5 | `arr[5] = 9` |

Resulting Index Array: `idx = [1, 3, 6, 0, 2, 4, 7, 5]`

### Searchsorted  
insert positions to keep order (binary search on sorted data)

```python
sorted_arr = np.array([1, 3, 5, 7, 9])
positions = np.searchsorted(sorted_arr, [2, 4, 8])
print(positions)
```

### `np.searchsorted()` Tabular Map

We have a pre-sorted array and we want to find where to insert the values `[2, 4, 8]` to keep it sorted.

1. The Sorted Array (`sorted_arr`)

| Index | 0 | 1 | 2 | 3 | 4 |
| :---: | :-: | :-: | :-: | :-: | :-: |
| Value | 1 | 3 | 5 | 7 | 9 |


2. Where Search Values Insert

| Target Value | Insertion Position | Visual Placement in Array | Resulting Index |
| :---: | :---: | :---: | :---: |
| 2 | Between `1` and `3` | `[1,` `[2]` `, 3, 5, 7, 9]` | 1 |
| 4 | Between `3` and `5` | `[1, 3,` `[4]` `, 5, 7, 9]` | 2 |
| 8 | Between `7` and `9` | `[1, 3, 5, 7,` `[8]` `, 9]` | 4 |

Result: `np.searchsorted(sorted_arr, [2, 4, 8])` outputs `[1, 2, 4]`.

## Counting  


```python
# nonzero elements matching a condition
arr = np.array([3, 1, 4, 1, 5, 9, 2, 6])

mask = arr > 4
print(np.count_nonzero(mask))
```

### `np.count_nonzero()` Tabular Map

How the condition `arr > 4` is evaluated element-by-element, and counted:

| Index | Original Value (`arr`) | Condition (`> 4`) | Mask Result (`mask`) | Contribution to Count |
| :---: | :---: | :---: | :---: | :---: |
| **0** | `3` | `3 > 4` | `False` | `0` (ignored) |
| **1** | `1` | `1 > 4` | `False` | `0` (ignored) |
| **2** | `4` | `4 > 4` | `False` | `0` (ignored) |
| **3** | `1` | `1 > 4` | `False` | `0` (ignored) |
| **4** | `5` | `5 > 4` | `True` | `1` (+1 to count) |
| **5** | `9` | `9 > 4` | `True` | `1` (+1 to count) |
| **6** | `2` | `2 > 4` | `False` | `0` (ignored) |
| **7** | `6` | `6 > 4` | `True` | `1` (+1 to count) |

**Total Non-Zero (True) Elements**: `1 + 1 + 1 = 3`

*Result:* `np.count_nonzero(mask)` outputs `3`.

```python
# Unique values and counts
arr = np.array([3, 1, 4, 1, 5, 9, 2, 6])

values, counts = np.unique(arr, return_counts=True)
print(values)
print(counts)
```

### `np.unique(..., return_counts=True)` Tabular Map

How the unique values and their frequencies are extracted from the original array:

1. The Original Array (`arr`)

| Index | 0 | 1 | 2 | 3 | 4 | 5 | 6 | 7 |
| :---: | :-: | :-: | :-: | :-: | :-: | :-: | :-: | :-: |
| Value | 3 | **1** | 4 | **1** | 5 | 9 | 2 | 6 |

*Note that `1` appears twice, while all other numbers appear exactly once.*

2. Unique Mapping Result

| Unique Value (`values`) | Occurrences in `arr` | Total Count (`counts`) |
| :---: | :---: | :---: |
| 1 | `arr[1]`, `arr[3]` | 2 |
| 2 | `arr[6]` | 1 |
| 3 | `arr[0]` | 1 |
| 4 | `arr[2]` | 1 |
| 5 | `arr[4]` | 1 |
| 6 | `arr[7]` | 1 |
| 9 | `arr[5]` | 1 |

* `values` = `[1, 2, 3, 4, 5, 6, 9]` (Sorted list of unique items)
* `counts` = `[2, 1, 1, 1, 1, 1, 1]` (Matching count of occurrences)

```python
# Logical combinations (element-wise)

a = np.array([True, False, True])
b = np.array([False, False, True])
print(np.logical_and(a, b))
print(np.logical_or(a, b))
```

### `np.logical_and()` & `np.logical_or()` Tabular Map

How the logical operations are evaluated element-by-element across the boolean arrays:

| Index | Element `a` | Element `b` | Logical AND (`a AND b`) | Logical OR (`a OR b`) |
| :---: | :---: | :---: | :---: | :---: |
| 0 | `True` | `False` | `False` | `True` |
| 1 | `False` | `False` | `False` | `False` |
| 2 | `True` | `True` | `True` | `True` |

* `np.logical_and(a, b)` outputs: `[False, False, True]`
* `np.logical_or(a, b)` outputs: `[True, False, True]`
