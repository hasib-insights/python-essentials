### Dictionary to Series & DataFrame

```python
#Create a Dictionary
members = {
    "Name": ["John", "Jane", "Jack", "Jill"],
    "Age": [26, 24, 22, 28],
    "City": ["Berlin", "London", "Moscow", "New York"]
}
```

```python
members
```

```python
#create a Series from the dictionary
s = pd.Series(members)
s
```

```python
#create a series from the dictionary with only the "Name" key
s = pd.Series(members["Name"])
s
```

```python
#create a DataFrame from the dictionary
df = pd.DataFrame(members)
df
```

### Set to Series and DataFrame

```python
set1 = {1, 2, 3, 4, 5}

print(set1)
type(set1)
```

```python
#create a series from the set
series = pd.Series(list(set1))
series
```

```python
#create a DataFrame from the set
df = pd.DataFrame(list(set1), columns=['Set_Values'])
df
```

### Tuple to Series and DataFrame

```python
t = (1, 2, 3, 4, 5, True)
print(t)
type(t)
```

```python
series = pd.Series(list(t))
series
```

```python
df = pd.DataFrame(list(t), columns=["Tuple_Values"])
df
```
