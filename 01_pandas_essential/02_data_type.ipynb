# Pandas Numeric Data Types

| **Data Type** | **Description**                                                       | **Common Pandas dtypes**          |
| ------------- | --------------------------------------------------------------------- | --------------------------------- |
| **Integer**   | Integer data types store whole numbers such as `-1`, `10`, and `100`. | `int8`, `int16`, `int32`, `int64` |
| **Float**     | Float data types store numbers that can contain decimal values.       | `float16`, `float32`, `float64`   |

---

# Pandas Integer Data Types — Range of Values

| **Data Type** | **Size (bits)** | **Size (bytes)** | **Range of Values**                                     | **Range Formula** |
| ------------- | --------------: | ---------------: | ------------------------------------------------------- | ----------------- |
| `int8`        |               8 |                1 | -128 to 127                                             | −2⁷ to 2⁷ − 1     |
| `int16`       |              16 |                2 | -32,768 to 32,767                                       | −2¹⁵ to 2¹⁵ − 1   |
| `int32`       |              32 |                4 | -2,147,483,648 to 2,147,483,647                         | −2³¹ to 2³¹ − 1   |
| `int64`       |              64 |                8 | -9,223,372,036,854,775,808 to 9,223,372,036,854,775,807 | −2⁶³ to 2⁶³ − 1   |

---

# Additional Pandas Data Types

| **Category**      | **Data Type** | **Description**                                                                                  | **Common Pandas dtype** |
| ----------------- | ------------- | ------------------------------------------------------------------------------------------------ | ----------------------- |
| **Date and Time** | Timestamp     | Represents dates and times.                                                                      | `datetime64[ns]`        |
| **Date and Time** | Timedelta     | Represents differences or durations between dates and times.                                     | `timedelta64[ns]`       |
| **Categorical**   | Category      | Represents data with a finite number of categories. Useful for **nominal** and **ordinal** data. | `category`              |
| **Text**          | String        | Represents text data.                                                                            | `string` / `object`     |
| **Boolean**       | Boolean       | Represents logical values such as **True** and **False**.                                        | `bool` / `boolean`      |


```python
import pandas as pd 
```

```python
#create a DataFrame with numeric data
data ={"age":[23, 45, 56, 25, 34, 65, 20, 30, 40, 45],
       "salary":[10000, 20000, 30000, 40000, 50000, 60000, 70000, 80000, 90000, 100000]}
df = pd.DataFrame(data)
df
```

```python
# Get the data types of each column in the Dataframe
df.dtypes
```

```python
# changing the data type of each column in the DataFrame 
df['age'] = df['age'].astype('int32')
df['salary'] = df['salary'].astype('float32')
df
```

```python
df.dtypes
```

### Datetime

```python
#create a list of dates
dates = ['2026', '2026', '2026', '2026', '2026']
#create a DataFrame with date data
df1 =pd.DataFrame(dates, columns=['date'])
df1
```

```python
df1.dtypes 
```

```python
#convert the data column to datetime format
timestamps = pd.to_datetime(df1['date'])
timestamps
```

```python
#create a DataFrame with the converted datetime data
df2 = pd.DataFrame({'dates': timestamps})
df2
```

```python
df2.dtypes
```

```python
#create a list of dates with time information
dates2 = ['2026-01-01 12:01:26.1215', '2026-02-01 12:02:25.5456', '2026-03-02 12:03:24.9897']

timestamps = pd.to_datetime(dates2)
df3 = pd.DataFrame({'dates': timestamps})
df3
```

```python
df3['dates'].dtypes
```
