### Reshape  
Change array dimensions without changing data:

````python
# Create 1D array by using arange() function
arr = np.arange(15)
arr
````

````python
# print the shape of the array
arr.shape
````

````python
# Reshape to 2D
arr2d = arr.reshape(3,5) # (row=3, column=5)
arr2d
````

````python
# varify the shape of the 2D array
arr2d.shape
````

````python
#reshape to 3d
arr3d = arr.reshape(3, 5, 1) # (row=3, column=5, depth=1)
arr3d
````

````python
# Use -1 for automatic dimension
arr3d_auto = arr.reshape(3, -1, 1) # (row=3, column=5, depth=1)
arr3d_auto
````

  ```
   Original arr has 15 elements.

   reshape(3, -1, 1)  →  3 × ? × 1  =  15

                         3 × ? × 1  =  15
                             ? = 15 / (3 × 1)
                             ? = 15 / 3
                             ? = 5  

   So NumPy replaces -1 with 5!
```
```
This will work: 12 elements = 3 × 4
arr.reshape(3, 4)  # works
This will fail: 12 elements ≠ 3 × 5
arr.reshape(3, 5)  # Error
```

# Resize  
Change array size (can add/remove elements):

````python
arr = np.array([1, 2, 3, 4, 5, 6])
# Resize to large (pads with zeros or repeats)
arr_resized = np.resize(arr, (3,3))
arr_resized
````

````python
# Resize to smaller (truncates)
arr_small = np.resize(arr, (2,2))
arr_small
````

```
Difference:

Reshape: Same number of elements, different shape
Resize: Can change number of elements
```

# Transpose
Swap rows and columns:

````python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print("Original:")
print(arr)
````

````python
# Transpose
arr_T = arr.T
print("\nTransposed:")
print(arr_T)
````

````python
# Or use transpose() method
arr_T2 = arr.transpose()
arr_T2
````

```
   (2, 3)                     (3, 2)
   ┌───┬───┬───┐              ┌───┬───┐
   │ 1 │ 2 │ 3 │              │ 1 │ 4 │
   ├───┼───┼───┤   transpose  ├───┼───┤
   │ 4 │ 5 │ 6 │   ────────>  │ 2 │ 5 │
   └───┴───┴───┘              ├───┼───┤
                              │ 3 │ 6 │
                              └───┴───┘

   2 rows, 3 cols             3 rows, 2 cols

# Flatten and Ravel
Both `flatten()` and `ravel()` convert a multi-dimensional array into a 1D array but they do it very differently under the hood!
```
flatten()  →  returns a COPY   (new memory)
ravel()    →  returns a VIEW   (shared memory)
```

````python
arr_2d = np.array([[1, 2, 3],
                   [3, 4, 5]])
# Flatten - creates a copy
flattened = arr_2d.flatten()
print(flattened)
````

````python
# Ravel - creates a view (if possible)
raveled = arr_2d.ravel()
print(raveled)
````

```
   ORIGINAL arr                      flatten() copy        ravel() view
   ┌─────────────────┐              ┌─────────────────┐   (points to arr)
   │  MEMORY 1000    │              │  MEMORY 2000    │
   │                 │  flatten()   │                 │
   │  [1,2,3,4,5,6]  │  ─────────>  │  [1,2,3,4,5,6]  │
   │                 │              │                 │
   └─────────────────┘              └─────────────────┘
          ▲
          │      ravel()
          └─────────────────┐              
                            │
                  (view shares memory)
```
---

````python
# Modify raveled (may affect original)
raveled[0] = 99
print(arr_2d)  # May or may not change depending on memory layout
````
